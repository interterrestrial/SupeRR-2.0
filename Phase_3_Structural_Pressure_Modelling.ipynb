# Phase 3 — Corrected Structural Pressure Modelling

This notebook is the authoritative executable entry point for corrected Phase 3. It consumes the validated Phase 2 master and intentionally does **not** reconstruct scores, targets, required rates, balls remaining, or legal-ball counts. The model uses pre-delivery structural state only.

## Locked methodology

- Exclude an entire match when any revised-overs or revised-target field is populated.
- Use the supplied source over annotation (1–20), not an inferred legal-ball counter.
- Construct current state from wickets already lost and wickets in prior delivery records; shift the current outcome out.
- Benchmark wickets leave-one-match-out within format × innings × source over, with predeclared broader fallbacks.
- Combine bounded wicket strain, prior-12-record wicket shock, and stage urgency with equal locked weights.
- Treat Low/Medium/High as descriptive within-format-and-innings percentile bands; retain the continuous index for modelling.
